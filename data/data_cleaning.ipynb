```python
df = data[['title', 'authors', 'description', 'average_rating', 'categories', 'published_year', 'ratings_count' ]]
df.head()
```

```python
print(df.shape)
```

```python
df = df.dropna(subset=['title'])
df['average_rating'] = df['average_rating'].fillna(0)
df['ratings_count'] = df['ratings_count'].fillna(0)
```

```python
df.info()
```

```python
missing_yrs = df[df['published_year'].isnull()]
print(missing_yrs[['title' , 'authors']])
```

```python
df.loc[df['title'] == 'Book Club', 'published_year'] = 2012
df.loc[df['title'] == 'The Civil War', 'published_year'] = 1958
df.loc[df['title'] == 'The Fellowship of the Ring', 'published_year'] = 1954
df.loc[df['title'] == 'Grade 4 Common Core Exemplar Collection', 'published_year'] = 2012
df.loc[df['title'] == 'Frankenstein: City of Night: A Novel', 'published_year'] = 2005
df.loc[df['title'] == '20000 LEAGUES UNDER THE SEA', 'published_year'] = 1870

df.loc[df['title'] == 'Frankenstein: City of Night: A Novel', 'authors'] = 'Dean Koontz'

df.info()
```

```python
df['authors'] = df['authors'].fillna("Unknown Author")
df['categories'] = df['categories'].fillna("Uncategorized")
df['description'] = df['description'].fillna("No description available.")

df.info()
```

Create a weighted rating for the books

```python
C = df['average_rating'].mean()
m = df['ratings_count'].quantile(0.90)

def calculate_weighted_rating(row, m=m, C=C):
    v = row['ratings_count']
    R = row['average_rating']
    if (v + m) == 0:
        return 0
    return (v/(v+m) * R) + (m/(m+v) * C)

df['weighted_rating'] = df.apply(calculate_weighted_rating, axis=1)

```

```python
df['context_string'] = (
    "Title: " + df['title'] + ". " +
    "Author(s): " + df['authors'] + ". " +
    "Category: " + df['categories'] + ". " +
    "Published: " + df['published_year'].astype(int).astype(str) + ". " +
    "Summary: " + df['description']
)

print("Sample Context String:\n")
print(df['context_string'].iloc[0])
print("\nTop 5 Books by Weighted Rating:")
print(df[['title', 'average_rating', 'ratings_count', 'weighted_rating']].sort_values(by='weighted_rating', ascending=False).head(5))
```
